# Difference Between Interpreter and Compiler

The main difference between an **interpreter** and a **compiler** lies in how they execute code.

| Feature         | Interpreter | Compiler |
|----------------|------------|----------|
| **Execution**  | Translates and runs code **line by line** | Translates the entire code **at once** before execution |
| **Speed**      | Slower, since it processes one line at a time | Faster, as it processes the whole code beforehand |
| **Error Handling** | Stops at the first error, making debugging easier | Shows all errors after compilation |
| **Output**     | Immediate, useful for testing and debugging | Runs only after full compilation |
| **Examples**   | Python, JavaScript, PHP | C, C++, Java (Java compiles to bytecode, then uses an interpreter) |

Since Python is an **interpreted** language, it executes code line by line, making debugging easier but slightly slower than compiled languages.


#  Variables and Comments in Python - Interview Questions & Answers

## **Variables in Python**

### **1. What are variables in Python? How do they differ from variables in other languages like C or Java?**
A **variable** in Python is a named reference to a value stored in memory. Unlike languages like C or Java, Python variables **do not require explicit type declaration**. Python is dynamically typed, meaning the variable type is inferred based on the assigned value.

```python
x = 10  # No need to declare int x;
name = "Alice"  # No need to declare String name;
```

---

### **2. How do you declare a variable in Python? Do you need to specify the data type?**
In Python, you declare a variable by assigning a value using the `=` operator. **You don’t need to specify the data type** because Python automatically determines it.

```python
age = 25       # Integer
price = 19.99  # Float
name = "John"  # String
is_valid = True  # Boolean
```

---

### **3. What are the different types of variables in Python?**
Python supports different data types, including:
- **Integer (`int`)** → `x = 10`
- **Floating Point (`float`)** → `y = 10.5`
- **String (`str`)** → `name = "Alice"`
- **Boolean (`bool`)** → `is_active = True`
- **List (`list`)** → `arr = [1, 2, 3]`
- **Tuple (`tuple`)** → `coords = (10, 20)`
- **Set (`set`)** → `unique_values = {1, 2, 3}`
- **Dictionary (`dict`)** → `person = {"name": "John", "age": 30}`

---

### **4. Explain the concept of dynamic typing in Python.**
Python is **dynamically typed**, meaning you don’t need to declare variable types explicitly. The type is determined at runtime based on the assigned value.

```python
x = 10      # Initially an integer
x = "hello" # Now a string (Python allows this)
```

---

### **5. What is the difference between mutable and immutable variables in Python? Give examples.**
- **Mutable** variables can be changed after assignment (e.g., lists, dictionaries, sets, bytearray).
- **Immutable** variables cannot be changed after assignment (e.g., integers, floats, strings, tuples, frozenset, bytes, Bool).

```python
# Mutable example
lst = [1, 2, 3]
lst.append(4)  # List is modified
print(lst)  # Output: [1, 2, 3, 4]

# Immutable example
name = "Alice"
name[0] = "B"  # ❌ This will cause an error because strings are immutable
```

---

### **6. How do global and local variables work in Python? Provide an example.**
- A **local variable** is defined inside a function and accessible only within that function.
- A **global variable** is defined outside any function and can be accessed anywhere in the script.

```python
x = 10  # Global variable

def my_function():
    x = 5  # Local variable
    print(x)  # Output: 5

my_function()
print(x)  # Output: 10 (global x is not changed)
```

---

### **7. What is the use of the `global` and `nonlocal` keywords in Python?**
- **`global`** is used to modify a global variable inside a function.
- **`nonlocal`** is used inside nested functions to modify a variable from an enclosing (non-global) scope.

```python
x = 10  

def modify_global():
    global x  
    x = 20  

modify_global()
print(x)  # Output: 20

# Using nonlocal
def outer():
    y = 5
    def inner():
        nonlocal y
        y = 10
    inner()
    print(y)  # Output: 10

outer()
```

---

### **8. What will happen if you use an undeclared variable in Python?**
You will get a `NameError`.

```python
print(x)  # NameError: name 'x' is not defined
```

---

### **9. What is variable shadowing in Python?**
Variable shadowing occurs when a **local variable** has the same name as a **global variable**, making the global variable inaccessible inside the function.

```python
x = 10  # Global variable

def func():
    x = 5  # This shadows the global x inside the function
    print(x)  # Output: 5

func()
print(x)  # Output: 10 (global variable remains unchanged)
```

---

### **10. How can you swap two variables in Python without using a third variable?**
```python
a, b = 5, 10
a, b = b, a  # Swapping in one line
print(a, b)  # Output: 10, 5
```

## **📌 Comments in Python**

### **1. What are comments in Python, and why are they important?**
Comments are used to explain code and make it more readable. They are ignored during execution.

---

### **2. How do you write a single-line comment in Python?**
Use the `#` symbol:

```python
# This is a single-line comment
x = 10  # This is also a comment
```

---

### **3. How do you write a multi-line comment in Python?**
Python doesn’t have a built-in multi-line comment syntax, but you can use triple quotes (`""" """` or `''' '''`).

```python
"""
This is a multi-line comment.
It spans multiple lines.
"""
```

---

### **4. What is a docstring in Python, and how is it different from regular comments?**
A **docstring** is a special type of multi-line comment used for documentation. Unlike normal comments, docstrings can be accessed using `__doc__`.

```python
def add(a, b):
    """This function adds two numbers."""
    return a + b

print(add.__doc__)  # Output: This function adds two numbers.
```

---

### **5. How can you retrieve a function’s docstring in Python?**
Using the `__doc__` attribute:

```python
print(add.__doc__)  # Retrieves the function's docstring
```

---

### **6. Can comments affect the execution of Python code?**
No, comments are ignored by the Python interpreter.

---

### **7. What is the recommended way to document a Python function or module?**
Using **docstrings**:

```python
def greet(name):
    """Returns a greeting message for the given name."""
    return f"Hello, {name}!"
```

---
### **8. Are comments stored in bytecode when Python compiles the script?**
No, comments are stripped out when Python compiles a script into bytecode (.pyc files).

---

### **9. Why is it a bad practice to leave unnecessary comments in production code?**
Too many comments can clutter code. Instead, write self-explanatory code and use comments only when necessary.

# Top 20 Interview Questions on Data Types, Input Function, and Data Conversion in Python

### Data Types

1. **What are the basic data types in Python?**  
   Python has several built-in data types including:  
   - `int`: Integer values like 1, 2, -3
   - `float`: Floating-point values like 3.14, -5.5
   - `str`: String values like "Hello", 'Python'
   - `list`: An ordered collection, e.g., [1, 2, 3]
   - `tuple`: An immutable ordered collection, e.g., (1, 2, 3)
   - `set`: An unordered collection of unique elements, e.g., {1, 2, 3}
   - `dict`: A collection of key-value pairs, e.g., {"name": "Alice", "age": 25}
   - `bool`: Boolean values, either `True` or `False`
   - `NoneType`: Represents the absence of a value, e.g., `None`.

2. **What is the difference between a list and a tuple in Python?**  
   - **List**: Mutable (can be changed after creation), allows duplicate elements, and supports various operations (e.g., append, remove).
   - **Tuple**: Immutable (cannot be modified after creation), typically used to store constant data, and allows duplicate elements.

3. **What is the difference between deep copy and shallow copy in Python?**  
   - **Shallow copy**: Creates a new object, but the elements within the object are still references to the original elements.
   - **Deep copy**: Creates a completely new object, and the elements within the object are also copied.

4. **What is the difference between `is` and `==` in Python?**  
   - `is`: Checks whether two variables point to the same object in memory.
   - `==`: Checks whether the values of two objects are equal.

5. **What are dictionaries in Python? How are they different from lists and tuples?**  
   - A **dictionary** is a collection of key-value pairs, where each key is unique. It provides fast access to its elements using keys.
   - A **list** and **tuple** are ordered collections, whereas a dictionary is unordered and does not allow duplicate keys.

6. **What is a set in Python? How is it different from a list?**  
   - A **set** is an unordered collection of unique elements, whereas a **list** is an ordered collection that can have duplicates.

7. **What is the purpose of the `None` data type in Python?**  
   - `None` represents the absence of a value or a null value. It is often used as a placeholder for optional or uninitialized variables.

8. **What is type casting in Python?**  
   - Type casting is the process of converting one data type into another. Python supports both implicit and explicit type casting (e.g., `int()`, `float()`, `str()` for explicit conversions).

9. **Can you explain the concept of variable type inference in Python?**  
   - Python automatically infers the data type of a variable when it is assigned a value, based on the type of the value being assigned.

10. **What are immutable data types in Python?**  
    - Immutable data types include `int`, `float`, `str`, `tuple`, and `frozenset`. These cannot be modified after their creation.
