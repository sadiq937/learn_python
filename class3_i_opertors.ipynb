# Python Interview Questions and Answers on Operators

## 1. What are Python operators?
**Answer:** Operators in Python are special symbols that perform operations on variables and values. Examples include `+`, `-`, `*`, `/`, etc.

---
## 2. What are the different types of operators in Python?
**Answer:** Python supports the following types of operators:
- **Arithmetic operators** (`+`, `-`, `*`, `/`, `//`, `%`, `**`)
- **Comparison operators** (`==`, `!=`, `>`, `<`, `>=`, `<=`)
- **Logical operators** (`and`, `or`, `not`)
- **Bitwise operators** (`&`, `|`, `^`, `~`, `<<`, `>>`)
- **Assignment operators** (`=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`)
- **Identity operators** (`is`, `is not`)
- **Membership operators** (`in`, `not in`)

---
## 3. What is the difference between `/` and `//` in Python?
**Answer:**
- `/` (Division Operator) returns a float result: `5 / 2 = 2.5`
- `//` (Floor Division Operator) returns an integer (floored) result: `5 // 2 = 2`

---
## 4. What is the modulus operator `%` used for?
**Answer:** The modulus operator `%` returns the remainder of a division:
```python
print(10 % 3)  # Output: 1
```

---
## 5. What is the exponentiation operator in Python?
**Answer:** The exponentiation operator `**` is used to raise a number to a power:
```python
print(2 ** 3)  # Output: 8
```

---
## 6. What is the difference between `is` and `==` operators?
**Answer:**
- `==` compares **values**:
  ```python
  a = [1, 2, 3]
  b = [1, 2, 3]
  print(a == b)  # True (values are same)
  ```
- `is` compares **memory locations**:
  ```python
  print(a is b)  # False (different objects in memory)
  ```

---
## 7. What are identity operators in Python?
**Answer:** Identity operators (`is`, `is not`) check if two objects refer to the same memory location.
```python
x = [1, 2, 3]
y = x
print(x is y)  # True
print(x is not y)  # False
```

---
## 8. How do `and`, `or`, and `not` logical operators work?
**Answer:**
- `and`: Returns `True` if both operands are `True`.
- `or`: Returns `True` if at least one operand is `True`.
- `not`: Reverses the boolean value.
```python
print(True and False)  # False
print(True or False)   # True
print(not True)        # False
```

---
## 9. What is the difference between `&` and `and` in Python?
**Answer:**
- `&` is a **bitwise AND** operator (works on bits).
- `and` is a **logical AND** operator (works on boolean values).
```python
print(5 & 3)   # 1 (Bitwise)
print(True and False)  # False (Logical)
```

---
## 10. What is the result of `3 | 1`?
**Answer:** `|` is the **bitwise OR** operator:
```python
print(3 | 1)  # Output: 3 (011 | 001 = 011)
```

---
## 11. What is the use of `^` operator?
**Answer:** `^` is the **bitwise XOR** operator:
```python
print(5 ^ 3)  # Output: 6 (101 ^ 011 = 110)
```

---
## 12. What is the bitwise NOT operator in Python?
**Answer:** The `~` operator inverts all bits of a number:
```python
print(~5)  # Output: -6 (Binary: ~(00000101) = 11111010)
```

---
## 13. What is the left shift (`<<`) operator?
**Answer:** Left shift moves bits to the left, filling with zeros.
```python
print(5 << 1)  # Output: 10 (Binary: 101 << 1 = 1010)
```

---
## 14. What is the right shift (`>>`) operator?
**Answer:** Right shift moves bits to the right.
```python
print(5 >> 1)  # Output: 2 (Binary: 101 >> 1 = 10)
```

---
## 15. What is the output of `bool(0)` and `bool(1)`?
**Answer:**
```python
print(bool(0))  # False
print(bool(1))  # True
```

---
## 16. How does the `+=` assignment operator work?
**Answer:** It increments and assigns in one step:
```python
x = 5
x += 3  # Same as x = x + 3
print(x)  # 8
```

---
## 17. What is the difference between `in` and `not in`?
**Answer:** These are **membership operators**:
```python
print(3 in [1, 2, 3])    # True
print(4 not in [1, 2, 3])  # True
```

---
## 18. What is operator precedence in Python?
**Answer:** Operator precedence determines the order in which operators are evaluated. Example:
```python
print(3 + 5 * 2)  # Output: 13 (* has higher precedence than +)
```

---
## 19. How can you use parentheses to change operator precedence?
**Answer:** Parentheses force evaluation order:
```python
print((3 + 5) * 2)  # Output: 16
```

---
## 20. What is short-circuiting in logical operators?
**Answer:** Python stops evaluating once the result is determined:
```python
print(False and print("Hello"))  # Does not print "Hello"
print(True or print("Hello"))   # Does not print "Hello"
```

---
## 21. What happens when we use `+` with strings?
**Answer:** String concatenation:
```python
print("Hello " + "World")  # Output: "Hello World"
```

---
## 22. Can we use `*` with strings?
**Answer:** Yes, for repetition:
```python
print("Hi" * 3)  # Output: "HiHiHi"
```

---
## 23. What is the result of `bool([])`?
**Answer:** Empty sequences are `False`:
```python
print(bool([]))  # False
```

---
## 24. Can you overload operators in Python?
**Answer:** Yes, using **magic methods** (`__add__`, `__sub__`, etc.).

---

x = 10

min_value = x if x < y else y
