# IF, Elif and Else Statements


````python
x = 125
y = 152
if x > y:
    print("x is greater than y")
elif x == y:
    print ("x is equal to y")
else :
    print("x is less than y")
````

####    Short Hand if and if-else

````python
x, y = 3, 2
if x >= y: print("this is short hand if") 
````

````python
x, y = 3, 2
print("this is short hand if") if x < y else print ("this is short hand ifelse")
````

#### These shorthand if statements are also called as *Ternary Operators*, or *Conditional Expressions*.

````python
marks = 85
print("A") if marks >= 90 else print("B") if marks >= 75 else print("C")
````

````python
x = 335
y = 315
print("x is less than y") if x < y else ("x is equal to y") if x == y else (" x is greater to y")
````

#### AND , OR , NOT Statements in IF Statements

````python
x = 125
y = 128
z = 130

if x < y and x < z:
    print("This is AND statement if both are true")
    
if x > y or x < z:
    print("This is OR statement if anyone is true")
    
if not x > y:
    print("This is Not statement when x is less than y")
````

### NESTED IF STATEMENTS

````python
x,y,z = 145,125,135

if x > y:
    print(" This is first loop")
    if x > z:
        print("     This is second loop")
        if x == z:
            print("x is equal to z")
        else:
            print("         This is third loop")
````

### Pass statement in IF
#### if statements cannot be empty, but if you for some reason have an if statement with no content, put in the pass statement to avoid getting an error.

````python
a,b,c = 130, 126, 128
if a  > b :
    pass
if a > b :
    print("Above if statement got pass due to pass statement")
````

# Python Interview Questions on Conditional and Transfer Statements

## **Conditional Statements**

### **1. What are conditional statements in Python?**
Conditional statements allow a program to execute different code blocks based on certain conditions using `if`, `elif`, and `else`.

### **2. Explain the syntax of an `if` statement in Python.**
```python
if condition:
    # Code block executed if condition is True
```
Example:
```python
age = 18
if age >= 18:
    print("You are eligible to vote.")
```

### **3. What is the difference between `if-elif-else` and multiple `if` statements?**
- `if-elif-else`: Only one condition is executed, even if multiple conditions are `True`.
- Multiple `if` statements: All conditions that evaluate to `True` will execute their corresponding blocks.

Example:
```python
x = 10
y = 20
if x > 5:
    print("X is greater than 5")
elif y > 15:
    print("Y is greater than 15")
else:
    print("Neither condition met")
```

### **4. Can an `if` statement exist without an `else`?**
Yes, an `if` statement does not require an `else`. The `else` block is optional.
```python
if 10 > 5:
    print("10 is greater than 5")
```

### **5. How can you write a single-line `if` statement?**
You can use the shorthand syntax:
```python
if 5 > 3: print("5 is greater than 3")
```

### **6. What is the ternary operator in Python?**
Python supports a compact `if-else` expression, also known as the ternary operator:
```python
x = 10
y = 20
min_value = x if x < y else y
print(min_value)  # Output: 10
```

## **Transfer Statements**

### **7. What are transfer statements in Python?**
Transfer statements alter the flow of execution. Python has `break`, `continue`, `pass`, and `return` statements.

### **8. What is the `break` statement in Python?**
The `break` statement terminates the loop prematurely.
```python
for i in range(5):
    if i == 3:
        break
    print(i)  # Output: 0 1 2
```

### **9. How does the `continue` statement work?**
The `continue` statement skips the current iteration and moves to the next one.
```python
for i in range(5):
    if i == 3:
        continue
    print(i)  # Output: 0 1 2 4
```

### **10. What is the `pass` statement?**
The `pass` statement acts as a placeholder and does nothing.
```python
for i in range(5):
    if i == 3:
        pass  # Placeholder for future implementation
    print(i)
```

### **11. What is the `return` statement used for?**
The `return` statement is used inside a function to return a value and exit the function.
```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8
```

### **12. What happens if a function has no `return` statement?**
If a function does not have a `return` statement, it returns `None` by default.
```python
def greet():
    print("Hello")

print(greet())  # Output: Hello \n None
```

## **Advanced Questions**

### **13. Can you use `else` with a loop?**
Yes, `else` can be used with `for` and `while` loops. It executes when the loop is completed normally (not terminated by `break`).
```python
for i in range(5):
    print(i)
else:
    print("Loop finished")  # Executes if no break occurs
```

### **14. How does `break` behave inside nested loops?**
A `break` statement only exits the loop in which it is used.
```python
for i in range(3):
    for j in range(3):
        if j == 1:
            break  # Breaks only inner loop
        print(i, j)
```

### **15. How do `break`, `continue`, and `pass` differ?**
| Statement | Functionality |
|-----------|--------------|
| `break`   | Exits the loop immediately |
| `continue`| Skips current iteration and moves to the next one |
| `pass`    | Placeholder that does nothing |
