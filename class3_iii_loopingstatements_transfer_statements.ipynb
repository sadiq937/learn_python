### Python Loops
Python has two primitive loop commands:

while loops
for loops


#### While Statement

````python
x = 25
y = 30
while x < y:
    print(x)
    x += 1
````

#### Note: remember to increment or decrement x or y, or else the loop will continue forever.

````python
x = 25
y = 30
while x < y:
    print(y)
    y -= 1
````

####  The break Statement
With the break statement we can stop the loop even if the while condition is true:

````python
x = 25
y = 30
while x < y:
    print(y)
    if y == 27:
        break
    y -= 1
````

#### Continue Statement
With the continue statement we can stop the current iteration, and continue with the next:

````python
x = 25
y = 30
while x < y:
    y -= 1
    if y == 27:
        continue
    print(y)
    
````

#### The else Statement
With the else statement we can run a block of code once when the condition no longer is true:

````python
x = 55
y = 47
while x >= y:
    print(y)
    y += 1
else:
    print("x is no longer greater than or equal to y")

    
````


# Top Python Interview Questions on Looping Statements

## 1. What is the difference between `for` loop and `while` loop in Python?
- A `for` loop is used for iterating over a sequence (like a list, tuple, dictionary, string, or range), and it runs for a fixed number of iterations.
- A `while` loop runs as long as a given condition is true, and the number of iterations may be unknown initially.

## 2. How can you iterate over a list in reverse order in Python?
You can reverse a list using the `reversed()` function or by using slicing with a step of `-1`.
```python
# Using reversed()
numbers = [1, 2, 3, 4]
for num in reversed(numbers):
    print(num)

# Using slicing
for num in numbers[::-1]:
    print(num)
```

## 3. What is the `break` statement in Python?
The `break` statement is used to exit a loop prematurely, terminating the loop's execution when a certain condition is met.
```python
for num in range(1, 10):
    if num == 5:
        break
    print(num)  # Output will be 1, 2, 3, 4
```

## 4. What is the `continue` statement in Python?
The `continue` statement is used to skip the current iteration of the loop and move to the next iteration.
```python
for num in range(1, 6):
    if num == 3:
        continue
    print(num)  # Output will be 1, 2, 4, 5
```

## 5. How can you iterate over a dictionary in Python?
You can iterate over a dictionary using `for` loops to access keys, values, or both.
```python
my_dict = {'a': 1, 'b': 2, 'c': 3}

# Iterating over keys
for key in my_dict:
    print(key)

# Iterating over values
for value in my_dict.values():
    print(value)

# Iterating over key-value pairs
for key, value in my_dict.items():
    print(key, value)
```

## 6. What is the `else` block in a loop in Python?
The `else` block in a loop is executed only if the loop completes without hitting a `break` statement.
```python
for num in range(1, 5):
    print(num)
else:
    print("Loop completed without break.")

# Output:
# 1
# 2
# 3
# 4
# Loop completed without break.
```

## 7. How do you loop through two lists simultaneously in Python?
You can use the `zip()` function to iterate over two or more lists at the same time.
```python
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']

for num, letter in zip(list1, list2):
    print(num, letter)
```

## 8. What is a nested loop in Python?
A nested loop is a loop inside another loop. The inner loop runs for each iteration of the outer loop.
```python
for i in range(3):
    for j in range(2):
        print(f"i = {i}, j = {j}")
```

## 9. How can you create an infinite loop in Python?
You can create an infinite loop using `while True:`. Be careful with such loops, as they will run indefinitely unless there is a `break` condition.
```python
while True:
    print("This is an infinite loop.")
    break  # Add break to stop the loop
```

## 10. How can you loop through a string in Python?
You can loop through each character in a string using a `for` loop.
```python
word = "Hello"
for char in word:
    print(char)
```
