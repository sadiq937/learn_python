````python
list1= [ "anju", "sadiq", "cherry", "watermelon"]

newlist1 = [ i for i in list1 if "a" in i]
print(newlist1)
````

##### The condition if "a" in i checks if the letter "a" is present in i. For 1, 34, -15.60, True, and [1, 2, 3, 4], this condition will raise a TypeError because these are not strings. For "apple", the condition will evaluate to True because "a" is present in the string.

##### Corrected Code: To avoid the TypeError, you can add a check to ensure that i is a string before applying the in operator. Here's the corrected code:

````python
my_list = [1, "apple", 34, -15.60, True, [1,2,3,4], "apple"]

new_list = [ i for i in my_list if isinstance(i, str) and"a" in i]

print(new_list)
````

# **Python List Methods - Interview Questions & Study Notes**

## **Study Notes on Lists in Python**

### **What is a List?**
A **list** in Python is an **ordered, mutable (modifiable) collection** of elements. It can contain **elements of different data types**, including numbers, strings, other lists, and even objects.

### **Key Properties of Lists:**
- Ordered (elements maintain their position)
- Mutable (can be modified after creation)
- Supports indexing and slicing
- Allows duplicate elements

### **Creating Lists:**
```python
# Empty List
my_list = []

# List with elements
my_list = [1, 2, 3, "apple", True]
```

### **Common List Operations:**
```python
# Accessing elements
print(my_list[0])   # First element
print(my_list[-1])  # Last element

# Slicing
print(my_list[1:3])  # Elements from index 1 to 2

# Modifying a List
my_list[2] = "banana"  # Change element at index 2
```

---

## **Python List Methods - Interview Questions**

### **1. How do you add an element to a list?**
```python
my_list.append(100)
```
- `append()` adds an element at the end of the list.

**Follow-up:** How is `append()` different from `extend()`?
```python
my_list.extend([200, 300])
```
- `extend()` adds multiple elements to the list, while `append()` adds only one.

---

### **2. What is the difference between `insert()` and `append()`?**
```python
my_list.insert(2, "Python")  # Inserts "Python" at index 2
```
- `insert(index, value)` inserts a value at a specific index, whereas `append()` adds to the end.

---

### **3. How do you remove an element from a list?**
```python
my_list.remove("apple")  # Removes first occurrence of "apple"
```
- `remove()` deletes the first occurrence of the specified value.
- If the value doesn’t exist, it raises an error.

**Follow-up:** How is `pop()` different?
```python
removed_item = my_list.pop(2)  # Removes and returns the element at index 2
```
- `pop(index)` removes an element at a specific index and returns it.

---

### **4. How do you reverse a list?**
```python
my_list.reverse()
```
- `reverse()` reverses the list **in place** (modifies the original list).
- You can also use slicing: `my_list[::-1]` (creates a new reversed list).

---

### **5. How do you sort a list in ascending and descending order?**
```python
my_list.sort()  # Ascending order
my_list.sort(reverse=True)  # Descending order
```
- `sort()` modifies the list in place.
- To get a sorted copy **without modifying the original**, use `sorted(my_list)`.

---

### **6. What is the difference between `sort()` and `sorted()`?**
- `sort()` modifies the list itself.
- `sorted()` returns a new sorted list without modifying the original.

```python
new_list = sorted(my_list)  # Creates a new sorted list
```

---

### **7. How do you find the index of an element in a list?**
```python
index = my_list.index("banana")
```
- `index(value)` returns the **first occurrence** of the element.
- If the element is not found, it raises an error.

---

### **8. How do you count occurrences of an element?**
```python
count = my_list.count("apple")
```
- `count(value)` returns the number of times an element appears in the list.

---

### **9. How do you copy a list?**
```python
new_list = my_list.copy()
```
- `copy()` creates a **shallow copy** of the list.
- You can also use `list(my_list)` or slicing `my_list[:]`.

---

### **10. What is the difference between `copy()` and `deepcopy()`?**
```python
import copy
new_list = copy.deepcopy(my_list)
```
- `copy()` creates a shallow copy (nested lists are still linked to the original).
- `deepcopy()` creates a completely independent copy.

---

### **11. How do you remove all elements from a list?**
```python
my_list.clear()
```
- `clear()` removes all elements, making the list empty (`[]`).

---

### **12. How do you iterate over a list?**
```python
for item in my_list:
    print(item)
```
- You can also use `enumerate()` to get the index:
```python
for index, item in enumerate(my_list):
    print(index, item)
```

---

### **13. How do you check if an element exists in a list?**
```python
if "apple" in my_list:
    print("Found!")
```

---

### **14. How do you concatenate two lists?**
```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
merged_list = list1 + list2
```
- You can also use `extend()`: `list1.extend(list2)`.

---

### **15. How do you find the maximum and minimum elements?**
```python
max_value = max(my_list)
min_value = min(my_list)
```
- Works only if the list contains comparable elements (all numbers or all strings).

---

## **Summary of List Methods**
| Method | Description |
|--------|-------------|
| `append(x)` | Adds an element at the end |
| `extend(iterable)` | Adds multiple elements |
| `insert(i, x)` | Inserts an element at index `i` |
| `remove(x)` | Removes first occurrence of `x` |
| `pop(i)` | Removes and returns element at index `i` |
| `clear()` | Removes all elements |
| `index(x)` | Finds index of `x` |
| `count(x)` | Counts occurrences of `x` |
| `sort()` | Sorts the list |
| `reverse()` | Reverses the list |
| `copy()` | Creates a shallow copy |
