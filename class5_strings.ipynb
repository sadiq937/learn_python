# **Python String Methods - Top Interview Questions**

## **Study Notes on Strings in Python**
### **1. What is a String in Python?**
A **string** is a sequence of characters enclosed in single (`'`), double (`"`), or triple quotes (`'''` or """).
```python
s1 = 'Hello'
s2 = "World"
s3 = '''Multiline String'''
```

### **2. Strings are Immutable**
Strings in Python **cannot be changed** after they are created.
```python
s = "hello"
s[0] = 'H'  # ❌ This will raise an error
```
To modify a string, you need to create a new one:
```python
s = "hello"
s = "H" + s[1:]  # "Hello"
```

### **3. Common String Methods**
| Method | Description |
|--------|-------------|
| `upper()` | Converts a string to uppercase |
| `lower()` | Converts a string to lowercase |
| `strip()` | Removes leading and trailing spaces |
| `replace(old, new)` | Replaces occurrences of a substring |
| `split(delimiter)` | Splits a string into a list based on a delimiter |
| `join(iterable)` | Joins elements of an iterable into a string |
| `find(substring)` | Finds the first occurrence of a substring |
| `startswith(prefix)` | Checks if the string starts with a given substring |
| `endswith(suffix)` | Checks if the string ends with a given substring |
| `isalpha()` | Checks if all characters are alphabets |
| `isdigit()` | Checks if all characters are digits |
| `isalnum()` | Checks if all characters are alphanumeric |

# **Top Python String Interview Questions**

## **1. How do you reverse a string in Python?**
```python
def reverse_string(s):
    return s[::-1]

print(reverse_string("hello"))  # Output: "olleh"
```

## **2. What is the difference between `find()` and `index()`?**
Both methods search for a substring but behave differently when the substring is not found:
```python
s = "hello world"
print(s.find("world"))  # Output: 6
print(s.find("Python"))  # Output: -1

print(s.index("world"))  # Output: 6
print(s.index("Python"))  # Raises ValueError ❌
```
**`find()` returns `-1`, while `index()` raises an error if the substring is not found.**

## **3. How do you check if a string contains only digits?**
```python
s = "12345"
print(s.isdigit())  # Output: True
```

## **4. How do you remove spaces from a string?**
```python
s = "  hello world  "
print(s.strip())   # Removes spaces from both ends
print(s.lstrip())  # Removes spaces from the left
print(s.rstrip())  # Removes spaces from the right
```

## **5. How do you count the occurrences of a character in a string?**
```python
s = "banana"
print(s.count("a"))  # Output: 3
```

## **6. How do you check if a string is a palindrome?**
```python
def is_palindrome(s):
    return s == s[::-1]

print(is_palindrome("radar"))  # Output: True
```

## **7. How do you convert a list of words into a sentence?**
```python
words = ["Python", "is", "awesome"]
print(" ".join(words))  # Output: "Python is awesome"
```

## **8. How do you check if a string starts or ends with a specific word?**
```python
s = "hello world"
print(s.startswith("hello"))  # Output: True
print(s.endswith("world"))    # Output: True
```

## **9. What is the difference between `split()` and `rsplit()`?**
```python
s = "apple,banana,grape,orange"
print(s.split(",", 2))   # Output: ['apple', 'banana', 'grape,orange']
print(s.rsplit(",", 2))  # Output: ['apple,banana', 'grape', 'orange']
```
**`split()` starts splitting from the left, while `rsplit()` starts from the right.**

## **10. How do you replace multiple occurrences of a substring?**
```python
s = "I love apples. Apples are tasty."
print(s.replace("apples", "bananas"))  # Case-sensitive replacement
```

## **11. How do you convert a string to a list of characters?**
```python
s = "hello"
print(list(s))  # Output: ['h', 'e', 'l', 'l', 'o']
```

## **12. How do you check if all characters in a string are lowercase?**
```python
s = "hello"
print(s.islower())  # Output: True
```

## **13. How do you find the ASCII value of a character?**
```python
print(ord("A"))  # Output: 65
```

## **14. How do you check if a string contains only whitespace?**
```python
s = "   "
print(s.isspace())  # Output: True
```

## **15. How do you format a string using placeholders?**
```python
name = "Alice"
age = 25
print(f"My name is {name} and I am {age} years old.")
```

- `isdigit()`, `isalpha()`, `isalnum()` are useful for validation.
- Strings can be **reversed**, **formatted**, and **searched efficiently** using built-in methods.
