# **Python Tuple - Interview Questions & Study Notes**

## **Tuple - Overview**
A **tuple** is an **immutable** sequence type in Python, used to store multiple items in a single variable. It is similar to a list, but it **cannot be modified (no add, remove, or replace operations).**

### **Key Features of Tuples:**
- **Immutable**: Cannot be changed after creation.
- **Ordered**: Maintains the order of elements.
- **Allows Duplicates**: Can store duplicate values.
- **Supports Indexing & Slicing**: Can be accessed using indices.
- **Faster than Lists**: Since tuples are immutable, they are optimized for speed.

### **Creating Tuples**
```python
# Empty tuple
empty_tuple = ()

# Tuple with elements
tuple1 = (1, 2, 3, 4)

# Tuple with mixed data types
tuple2 = ("hello", 10, 3.14, True)

# Tuple with a single element (comma needed)
single_tuple = (5,)
```

## **Common Tuple Methods**
Tuples have limited built-in methods due to immutability. Here are the most used ones:

| Method | Description | Example |
|--------|------------|---------|
| `count(value)` | Returns the number of times a value appears in a tuple | `t.count(3)` |
| `index(value, start, end)` | Returns the first index of a value | `t.index(3)` |

### **Tuple Operations**
Tuples support basic operations like:
```python
# Concatenation
t1 = (1, 2)
t2 = (3, 4)
print(t1 + t2)  # (1, 2, 3, 4)

# Repetition
print(t1 * 3)  # (1, 2, 1, 2, 1, 2)

# Membership Test
print(2 in t1)  # True

# Slicing
tuple3 = (10, 20, 30, 40, 50)
print(tuple3[1:4])  # (20, 30, 40)
```

## **Top Python Interview Questions on Tuple Methods**

### **1. What is a tuple in Python? How is it different from a list?**
A tuple is an **immutable** sequence of elements, whereas a list is **mutable**. Tuples use `()` while lists use `[]`.

### **2. Can you modify a tuple after creation?**
No, tuples are immutable. You cannot change, add, or remove elements after creation.

### **3. How can you access elements in a tuple?**
By using **indexing and slicing**:
```python
t = (10, 20, 30, 40)
print(t[1])  # 20
print(t[1:3])  # (20, 30)
```

### **4. How do you find the index of an element in a tuple?**
Using the `index()` method:
```python
t = (10, 20, 30, 40)
print(t.index(30))  # 2
```

### **5. How do you count occurrences of an element in a tuple?**
Using the `count()` method:
```python
t = (1, 2, 3, 1, 2, 1)
print(t.count(1))  # 3
```

### **6. How do you check if an element exists in a tuple?**
Using the `in` keyword:
```python
t = (10, 20, 30)
print(20 in t)  # True
```

### **7. How can you convert a tuple into a list?**
Using `list()`:
```python
t = (10, 20, 30)
l = list(t)
print(l)  # [10, 20, 30]
```

### **8. How can you convert a list into a tuple?**
Using `tuple()`:
```python
l = [10, 20, 30]
t = tuple(l)
print(t)  # (10, 20, 30)
```

### **9. How do you unpack a tuple?**
Tuple unpacking allows assigning tuple elements to variables:
```python
t = (10, 20, 30)
a, b, c = t
print(a, b, c)  # 10 20 30
```

### **10. Can a tuple contain other tuples?**
Yes, tuples can be nested:
```python
t = (1, (2, 3), (4, 5))
print(t[1])  # (2, 3)
```

### **11. How do you create a tuple with a single element?**
By adding a trailing comma:
```python

```

### **12. What happens if you try to modify a tuple?**
It will raise an error:
```python

t[0] = 5  # TypeError: 'tuple' object does not support item assignment
```

### **13. How can you merge two tuples?**
Using concatenation (`+`):
```python

t2 = (3, 4)
print(t1 + t2)  # (1, 2, 3, 4)
```

### **14. How can you repeat elements in a tuple?**
Using the `*` operator:
```python
t = (1, 2)
print(t * 3)  # (1, 2, 1, 2, 1, 2)
```

### **15. Are tuples more memory-efficient than lists?**
Yes, because tuples are immutable, they use **less memory** and are **faster** than lists.

### **16. How do you compare two tuples?**
Using comparison operators (`==`, `<`, `>`, etc.). Tuples are compared **element by element**:
```python

```

### **17. Can tuples have mutable objects?**
Yes, tuples can contain lists:

### **18. How do you sort a tuple?**
Convert it to a list, sort, and convert back:

### **19. How do you find the maximum and minimum in a tuple?**
Using `max()` and `min()`:

Unpacking:

````python
# Sorting in Tuples: `sort()` vs `sorted()`

"""
In Python, `sort` and `sorted` are used to sort elements in a collection. However, they behave differently, especially when working with tuples, which are immutable.
"""

# =========================================
# 1. `sorted()` Function
# =========================================

"""
The `sorted()` function returns a new sorted list from the elements of any iterable (like a tuple). It does not modify the original tuple.
"""

# Syntax:
# sorted(iterable, key=None, reverse=False)

# - `iterable`: The collection to sort (e.g., a tuple).
# - `key`: A function to customize sorting (optional).
# - `reverse`: If `True`, sorts in descending order (default is `False`).

# Example with Tuple:
my_tuple = (3, 1, 4, 1, 5, 9, 2)

# Sort the tuple and return a new list
sorted_list = sorted(my_tuple)

print("Original tuple:", my_tuple)  # Original tuple remains unchanged
print("Sorted list:", sorted_list)  # New sorted list

# Output:
"""
Original tuple: (3, 1, 4, 1, 5, 9, 2)
Sorted list: [1, 1, 2, 3, 4, 5, 9]
"""

# =========================================
# 2. `sort()` Method
# =========================================

"""
The `sort()` method is a list method that sorts the list in place (modifies the original list). It does not work directly with tuples because tuples are immutable.
"""

# Syntax:
# list.sort(key=None, reverse=False)

# - `key`: A function to customize sorting (optional).
# - `reverse`: If `True`, sorts in descending order (default is `False`).

# Example with List (Cannot Use `sort()` on Tuples):
my_list = [3, 1, 4, 1, 5, 9, 2]

# Sort the list in place
my_list.sort()

print("Sorted list:", my_list)

# Output:
"""
Sorted list: [1, 1, 2, 3, 4, 5, 9]
"""

# =========================================
# 3. Sorting a Tuple
# =========================================

"""
Since tuples are immutable, you cannot use the `sort()` method directly on them. Instead, you can:
1. Use `sorted()` to get a sorted list.
2. Convert the sorted list back to a tuple if needed.
"""

# Example:
my_tuple = (3, 1, 4, 1, 5, 9, 2)

# Sort the tuple and convert the result back to a tuple
sorted_tuple = tuple(sorted(my_tuple))

print("Original tuple:", my_tuple)
print("Sorted tuple:", sorted_tuple)

# Output:
"""
Original tuple: (3, 1, 4, 1, 5, 9, 2)
Sorted tuple: (1, 1, 2, 3, 4, 5, 9)
"""

# =========================================
# 4. Sorting with Custom Key
# =========================================

"""
You can use the `key` parameter in `sorted()` or `sort()` to customize the sorting logic.
"""

# Example:
# Sort a tuple of strings based on their length:
my_tuple = ("apple", "banana", "cherry", "date")

# Sort by length of the strings
sorted_tuple = tuple(sorted(my_tuple, key=len))

print("Sorted by length:", sorted_tuple)

# Output:
"""
Sorted by length: ('date', 'apple', 'cherry', 'banana')
"""

# =========================================
# 5. Sorting in Descending Order
# =========================================

"""
Use the `reverse=True` parameter to sort in descending order.
"""

# Example:
my_tuple = (3, 1, 4, 1, 5, 9, 2)

# Sort in descending order
sorted_tuple = tuple(sorted(my_tuple, reverse=True))

print("Sorted in descending order:", sorted_tuple)

# Output:
"""
Sorted in descending order: (9, 5, 4, 3, 2, 1, 1)
"""

# =========================================
# Key Differences Between `sort()` and `sorted()`
# =========================================

"""
| Feature              | `sort()`                          | `sorted()`                        |
|----------------------|-----------------------------------|-----------------------------------|
| Works on             | Lists only                       | Any iterable (tuples, lists, etc.) |
| Modifies original    | Yes (sorts in place)             | No (returns a new sorted list)    |
| Return value         | `None` (modifies the list)       | New sorted list                   |
| Immutable collections| Cannot be used (e.g., tuples)    | Can be used                       |
"""

# =========================================
# Summary
# =========================================

"""
- Use `sorted()` for tuples or when you want a new sorted list without modifying the original.
- Use `sort()` for lists when you want to sort in place.
- Tuples are immutable, so you cannot use `sort()` directly on them.
"""
````
