# **Python Dictionary: Study Notes & Interview Questions**

## **📌 Introduction to Dictionary in Python**
A **dictionary** in Python is an **ordered**, **mutable**, and **indexed** collection that stores key-value pairs.

### **✅ Key Features of Dictionaries:**
- **Keys must be unique** and **immutable** (e.g., strings, numbers, tuples).
- **Values can be mutable** (lists, dictionaries, etc.).
- **Dictionary items are unordered** (as of Python 3.7+, they maintain insertion order).
- **Access time complexity is O(1)** on average due to hash mapping.

### **✅ Creating a Dictionary:**
```python
# Using curly braces
my_dict = {"name": "Alice", "age": 25, "city": "New York"}

# Using dict() function
my_dict2 = dict(name="Bob", age=30, city="Los Angeles")
```

### **✅ Accessing Values in Dictionary:**
```python
print(my_dict["name"])  # Alice
print(my_dict.get("age"))  # 25
```

# **📌 Top Dictionary Methods Interview Questions**

## **1️⃣ How do you update an existing dictionary in Python?**
```python
my_dict["age"] = 26  # Updates value
my_dict.update({"city": "San Francisco", "job": "Engineer"})
```
**👉 Explanation:**
- Using `my_dict[key] = value` updates a single key.
- Using `.update()` updates multiple key-value pairs.

---

## **2️⃣ What is the difference between `.get()` and `[]` for retrieving values?**
```python
print(my_dict.get("salary"))  # None (No error if key doesn’t exist)
print(my_dict["salary"])  # KeyError if key doesn’t exist
```
**👉 Explanation:**
- `.get()` **returns `None`** (or a default value) if the key doesn’t exist.
- `[]` **raises `KeyError`** if the key doesn’t exist.

---

## **3️⃣ How do you remove a key from a dictionary?**
```python
my_dict.pop("age")  # Removes and returns the value
```
**👉 Explanation:**
- `.pop(key)` removes the key and returns its value.
- `.popitem()` removes and returns the **last inserted item** (Python 3.7+).
- `del my_dict["age"]` deletes the key without returning it.

---

## **4️⃣ How do you iterate through a dictionary?**
```python
# Iterate over keys
for key in my_dict:
    print(key, my_dict[key])

# Iterate over key-value pairs
for key, value in my_dict.items():
    print(key, value)
```
**👉 Explanation:**
- `.keys()` retrieves keys.
- `.values()` retrieves values.
- `.items()` retrieves key-value pairs.

---

## **5️⃣ How do you merge two dictionaries?**
```python
# Method 1: Using update()
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
dict1.update(dict2)

# Method 2: Using dictionary unpacking (Python 3.5+)
merged_dict = {**dict1, **dict2}
```
**👉 Explanation:**
- `.update()` modifies the original dictionary.
- `**` (dictionary unpacking) creates a new merged dictionary.

---

## **6️⃣ How do you create a dictionary with default values?**
```python
keys = ["name", "age", "city"]
default_dict = dict.fromkeys(keys, "Unknown")
```
**👉 Explanation:**
- `dict.fromkeys(keys, default_value)` creates a dictionary with default values.

---

## **7️⃣ How do you check if a key exists in a dictionary?**
```python
if "age" in my_dict:
    print("Key exists")
```
**👉 Explanation:**
- The `in` keyword checks for key existence efficiently.

---

## **8️⃣ What is the difference between `pop()` and `popitem()`?**
```python
print(my_dict.pop("name"))  # Removes and returns "name" value
print(my_dict.popitem())  # Removes and returns the last inserted item (Python 3.7+)
```
**👉 Explanation:**
- `pop(key)` removes a specific key.
- `popitem()` removes the last inserted key-value pair.

---

## **9️⃣ How do you copy a dictionary?**
```python
copy_dict = my_dict.copy()
```
**👉 Explanation:**
- `.copy()` creates a **shallow copy** of the dictionary.
- `dict(original_dict)` is another way to copy.

---

## **🔟 What is dictionary comprehension?**
```python
squared_numbers = {x: x**2 for x in range(5)}
```
**👉 Explanation:**
- `{key: value for key in iterable}` is dictionary comprehension.
- Example: `{x: x**2 for x in range(5)}` creates `{0: 0, 1: 1, 2: 4, 3: 9, 4: 16}`.

---

# **📌 Summary of Important Dictionary Methods**

| Method | Description |
|--------|------------|
| `dict.get(key, default)` | Returns value for key; avoids KeyError |
| `dict.update({key: value})` | Updates dictionary with key-value pairs |
| `dict.pop(key)` | Removes a key and returns its value |
| `dict.popitem()` | Removes and returns the last key-value pair |
| `dict.keys()` | Returns all keys |
| `dict.values()` | Returns all values |
| `dict.items()` | Returns all key-value pairs |
| `dict.copy()` | Returns a shallow copy of dictionary |
| `dict.clear()` | Removes all items from dictionary |
| `dict.setdefault(key, default)` | Returns value if key exists, otherwise sets it to default |
| `dict.fromkeys(iterable, default)` | Creates a dictionary with given keys and default value |
