# **Python Interview Questions on Sets and Frozensets**

## **📌 Study Notes: Sets in Python**
### **What is a Set?**
A **set** in Python is an **unordered**, **mutable**, and **unique collection** of elements. It is useful for **removing duplicates** and performing **set operations** like union, intersection, and difference.

### **Key Features of Sets**
- **Unordered**: The elements have no specific order.
- **Mutable**: We can add or remove elements.
- **Unique**: No duplicate elements allowed.
- **Uses Hashing**: Optimized for membership testing (`in` keyword).

### **Common Set Methods**
| Method | Description |
|--------|-------------|
| `add(x)` | Adds an element `x` to the set. |
| `remove(x)` | Removes `x` from the set (throws an error if not found). |
| `discard(x)` | Removes `x` without error if not found. |
| `pop()` | Removes and returns a random element. |
| `clear()` | Removes all elements. |
| `union(set2)` | Returns a new set with elements from both sets. |
| `intersection(set2)` | Returns common elements. |
| `difference(set2)` | Returns elements only in the first set. |
| `symmetric_difference(set2)` | Returns elements not in both sets. |
| `issubset(set2)` | Checks if set is a subset. |
| `issuperset(set2)` | Checks if set is a superset. |

---

## **📌 Study Notes: Frozensets in Python**
### **What is a Frozenset?**
A **frozenset** is an **immutable** version of a set. Once created, elements **cannot** be added or removed.

### **Key Features of Frozensets**
- **Immutable**: Cannot be modified after creation.
- **Supports Set Operations**: Can perform `union()`, `intersection()`, etc.
- **Useful as Dictionary Keys**: Since frozensets are hashable, they can be used as dictionary keys.

### **Common Frozenset Methods**
| Method | Description |
|--------|-------------|
| `union(set2)` | Returns a new frozenset with elements from both sets. |
| `intersection(set2)` | Returns common elements. |
| `difference(set2)` | Returns elements only in the first set. |
| `symmetric_difference(set2)` | Returns elements not in both sets. |
| `issubset(set2)` | Checks if frozenset is a subset. |
| `issuperset(set2)` | Checks if frozenset is a superset. |

# **🔥 Top Python Interview Questions on Sets & Frozensets**

### **1️⃣ How do you create a set in Python?**
```python
my_set = {1, 2, 3, 4}
print(my_set)  # Output: {1, 2, 3, 4}
```

### **2️⃣ How do you create an empty set?**
```python
empty_set = set()  # NOT {}
print(type(empty_set))  # Output: <class 'set'>
```

### **3️⃣ How do you create a frozenset?**
```python
frozen = frozenset([1, 2, 3, 4])
print(frozen)  # Output: frozenset({1, 2, 3, 4})
```

### **4️⃣ What happens if you try to add an element to a frozenset?**
```python
frozen.add(5)  # AttributeError: 'frozenset' object has no attribute 'add'
```

### **5️⃣ How do you check if an element exists in a set?**
```python
my_set = {1, 2, 3}
print(2 in my_set)  # Output: True
```

### **6️⃣ How do you remove an element from a set without causing an error if it doesn’t exist?**
```python
my_set.discard(10)  # No error even if 10 is not in the set
```

### **7️⃣ What is the difference between `remove()` and `discard()`?**
```python
s = {1, 2, 3}
s.remove(4)  # KeyError: 4 not in set
s.discard(4)  # No error
```

### **8️⃣ How do you find the union of two sets?**
```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1 | set2)  # Output: {1, 2, 3, 4, 5}
```

### **9️⃣ How do you find the intersection of two sets?**
```python
print(set1 & set2)  # Output: {3}
```

### **🔟 What is the difference between `difference()` and `symmetric_difference()`?**
```python
print(set1 - set2)  # Output: {1, 2} (Only in set1)
print(set1 ^ set2)  # Output: {1, 2, 4, 5} (Unique to each set)
```

### **1️⃣1️⃣ Can you use a set as a dictionary key?**
```python
my_dict = {{1, 2, 3}: "value"}  # TypeError: unhashable type: 'set'
```
❌ **Sets are mutable and unhashable, so they CANNOT be dictionary keys.**

### **1️⃣2️⃣ Can you use a frozenset as a dictionary key?**
```python
my_dict = {frozenset({1, 2, 3}): "value"}
print(my_dict)  # ✅ Works fine
```
✅ **Frozensets are immutable and hashable, so they CAN be dictionary keys.**

### **1️⃣3️⃣ What is the time complexity of checking an element in a set?**
✅ **O(1) (average case) because sets use hashing.**

### **1️⃣4️⃣ How do you convert a list to a set to remove duplicates?**
```python
numbers = [1, 2, 2, 3, 4, 4]
unique_numbers = set(numbers)
print(unique_numbers)  # Output: {1, 2, 3, 4}
```

### **1️⃣5️⃣ Can a set contain another set?**
❌ No, because sets are mutable and unhashable.
```python
s = {{1, 2, 3}}  # TypeError: unhashable type: 'set'
```

### **1️⃣6️⃣ Can a set contain a frozenset?**
✅ Yes, because frozensets are immutable.
```python
s = {frozenset({1, 2, 3})}  # ✅ Works fine
```

- **Use frozensets** when you need an immutable set.

#### **Accessing items in set**

````python
# Looping through set
my_set = {1, "apple", (52,85,94), False}

for i in my_set:
    print(i)

i = 0
while i < len(my_set):
    print(i)
    i += 1
````

````python
## Membership operators
my_set = {1, "apple", (52,85,94), False}

print("apple" in my_set)
print("banana" not in my_set)
````

#### **to add a single item in set we use add, for adding multiple items or iterables we use update**

````python
my_set = {1, "apple", (52,85,94), False}

my_set.add("sadiq")

print(my_set)
````

````python
my_set = {1, "apple", (52,85,94), False}
new_set = [1, 5, 84,"anju"]
my_set.update(new_set)

print(my_set)
````

#### **TO remove a item from in a set, we use remove() or discard**

````python
my_set = {1, "apple", (52,85,94), False}

# my_set.remove("cherry") # As cherry is not there in set, it raises error, unlike in discard

my_set.discard("cherry") # As cherry is not there in set, using discard wont raise error

print(my_set)
````

````python
my_set = {1, "apple", (52,85,94), False,25}

my_set.pop() # As sets doesn't have index so pop removes any random item

print(my_set)
````

````python
my_set = {1, "apple", (52,85,94), False}

for i in my_set:
    print(i)
````

#### **The update() method inserts all items from one set into another.**

#### **The update() changes the original set, and does not return a new set.**

````python
my_set = {1, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56, 23, 31}

my_set.update(my_set1)

print(my_set)
````

#### **The union() method returns a new set with all items from both sets, You can add list,tuple to set using this**

````python
my_set = {1, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56, 23, 31}
my_set2 = ["a","b"]
my_set3 = ("c","d")

new_set = my_set.union(my_set1, my_set2, my_set3)
print(new_set)
````

````python
my_set = {1, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56, 23, 31}

new_set = my_set | my_set1  # This union symbol can only be used to join sets with sets only

print(new_set)
````

#### **The intersection() method will return a new set, that only contains the items that are present in both sets.**

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = (78, 77 ,56,1, 23, 31) ## we intersect tuple,list with set

new = my_set.intersection(my_set1)

print(new)
````

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56,1, 23, 31}       
''' Note: The & operator only allows you to join sets with sets, 
and not with other data types like you 
can with the intersection() method.'''

new = my_set & my_set1

print(new)
````

#### **The intersection_update() method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set. - similar for all update method , it will change the original set**

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = (78, 77 ,56,1, 23, 31, 0) 

my_set.intersection_update(my_set1)

print(my_set)
````

#### **The difference() method will return a new set that will contain only the items from the first set that are not present in the other set.**

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = (78, 77 ,56,1, 23, 31, 0)

new_set = my_set.difference(my_set1)

print(new_set)
````

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56,1, 23, 31, 0}     ## The - operator only allows you to join sets with sets, and not with other data types like you can with the difference() method.

new_set = my_set - my_set1

print(new_set)
````

#### **The difference_update() method will also keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.**

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = (78, 77 ,56,1, 23, 31, 0) 

my_set.difference_update(my_set1)

print(my_set)
````

#### **The symmetric_difference() method will keep only the elements that are NOT present in both sets.**

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56,1, 23, 31, 0}

new_set = my_set.symmetric_difference(my_set1)

print(new_set)
````

````python
''' The ^ operator only allows you to join sets with sets,
and not with other data types like you can with the symmetric_difference() method'''
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = {78, 77 ,56,1, 23, 31, 0}

new_set = my_set^my_set1

print(new_set)
````

#### **The symmetric_difference_update() method will also keep all but the duplicates, but it will change the original set instead of returning a new set.**

````python
my_set = {1,78, "apple", (52,85,94), False}
my_set1 = (78, 77 ,56,1, 23, 31, 0) 

my_set.symmetric_difference_update(my_set1)

print(my_set)
````
