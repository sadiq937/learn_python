#### **It is best practise to using With to open a file beacuse there is no need to give close() seperately as it automatically closes file after use, saving resources and data integrety**

#### When you open a file using `file = open("text2.txt", "w")`, Python creates a connection (or file handle) between your program and the file on the disk. This connection consumes system resources, and if not properly closed, it can lead to:

- **Resource Leaks**: The file handle remains open, consuming memory and system resources.
- **Data Loss**: If the program crashes or exits unexpectedly, any buffered data (data not yet written to the file) may be lost.
- **File Locking**: The file may remain locked, preventing other programs or processes from accessing it.

#### Using `file.close()` ensures:
- All data is properly written to the file.
- The file handle is released, freeing up system resources.
- The file is unlocked, allowing other programs to access it.

````python
## using r+

with open("text1.txt", "r+") as x:
    print(x.tell()) # This will tell where the file pointer is located
    print(x.read())
    print(x.tell()) # after reading the file the file pointer has changed the position to end
    
    
````

````python
##  using w+ and a+

with open("text3.txt", "w+") as fd:
    print(fd.tell())
    x = fd.write("This is write function")
    print(fd.tell())
    fd.seek(0)  # this will move file pointer to starting as we gave 0
    y =fd.read()
    print(y)
    
    
with open("text3.txt", "a+") as fd:
    fd.write("This is doing append")
    fd.seek(0)
    z = fd.read()
    print(z)
    
````

| **Mode** | **Description**                                                                 | **File Existence** | **Pointer Position** | **Behavior**                                                                 |
|----------|---------------------------------------------------------------------------------|--------------------|----------------------|------------------------------------------------------------------------------|
| `r`      | **Read mode**                                                                   | Must exist         | Beginning            | Opens the file for reading. Raises `FileNotFoundError` if the file doesn't exist. |
| `w`      | **Write mode**                                                                  | Creates if missing | Beginning            | Opens the file for writing. Overwrites the file if it exists.               |
| `a`      | **Append mode**                                                                 | Creates if missing | End                  | Opens the file for appending. Adds data to the end of the file.             |
| `r+`     | **Read and write mode**                                                         | Must exist         | Beginning            | Opens the file for both reading and writing. Does not truncate the file.    |
| `w+`     | **Write and read mode**                                                         | Creates if missing | Beginning            | Opens the file for both writing and reading. Overwrites the file if it exists. |
| `a+`     | **Append and read mode**                                                        | Creates if missing | End                  | Opens the file for both appending and reading. Adds data to the end of the file. |

---

### **Key Notes:**
- **`r`**: Only allows reading. File must exist.
- **`w`**: Only allows writing. Overwrites the file if it exists.
- **`a`**: Only allows appending. Preserves existing content.
- **`r+`**: Allows both reading and writing. File must exist. Does not truncate the file.
- **`w+`**: Allows both writing and reading. Overwrites the file if it exists.
- **`a+`**: Allows both appending and reading. Preserves existing content.

---

### **Example Use Cases:**
| **Mode** | **Use Case**                                                                 |
|----------|------------------------------------------------------------------------------|
| `r`      | Reading data from an existing file.                                         |
| `w`      | Creating a new file or overwriting an existing file.                        |
| `a`      | Adding new data to the end of an existing file (e.g., logging).             |
| `r+`     | Reading and modifying an existing file without truncating it.               |
| `w+`     | Creating a new file or overwriting an existing file, then reading from it.  |
| `a+`     | Appending data to a file and reading from it.                               |

# File Handling in Python - Interview Questions & Notes

## 📌 Notes on File Handling in Python

### 1. Opening a File
```python
file = open('example.txt', 'r')  # Opens file in read mode
```
Modes:
- `'r'` – Read (default)
- `'w'` – Write (overwrites if file exists)
- `'a'` – Append (adds content at the end)
- `'x'` – Exclusive creation (fails if file exists)
- `'b'` – Binary mode (`'rb'`, `'wb'`)
- `'t'` – Text mode (default)

### 2. Reading a File
```python
with open('example.txt', 'r') as file:
    content = file.read()  # Reads entire file
    lines = file.readlines()  # Reads all lines into a list
```

### 3. Writing to a File
```python
with open('example.txt', 'w') as file:
    file.write('Hello, World!\n')  # Overwrites file
```

### 4. Appending to a File
```python
with open('example.txt', 'a') as file:
    file.write('Appended text.\n')
```

### 5. Working with Binary Files
```python
with open('image.jpg', 'rb') as file:
    data = file.read()
```

### 6. Checking if File Exists
```python
import os
if os.path.exists('example.txt'):
    print('File exists')
```

## ❓ Top Interview Questions on File Handling

### 1️⃣ What is file handling in Python?
**Answer:** File handling in Python allows reading, writing, and manipulating files on the system using built-in functions like `open()`, `read()`, `write()`, and `close()`.

---
### 2️⃣ What is the difference between `w` and `a` mode?
**Answer:**
- `'w'` mode overwrites the file if it exists or creates a new file.
- `'a'` mode appends data to the file without deleting existing content.

---
### 3️⃣ How do you read only the first 5 lines of a file?
```python
with open('example.txt', 'r') as file:
    for _ in range(5):
        print(file.readline().strip())
```

---
### 4️⃣ What is the use of `with` statement in file handling?
**Answer:** The `with` statement automatically closes the file after execution, preventing memory leaks and file corruption.
```python
with open('example.txt', 'r') as file:
    data = file.read()
```

---
### 5️⃣ How to copy contents of one file to another in Python?
```python
with open('source.txt', 'r') as src, open('destination.txt', 'w') as dest:
    dest.write(src.read())
```

---
### 6️⃣ How do you handle file exceptions in Python?
```python
try:
    with open('example.txt', 'r') as file:
        data = file.read()
except FileNotFoundError:
    print('File not found')
except IOError:
    print('I/O error occurred')
```

---
### 7️⃣ How to get the current file position while reading?
```python
with open('example.txt', 'r') as file:
    print(file.tell())  # Returns current position
    file.read(5)  # Reads 5 characters
    print(file.tell())  # Updated position
```

---
### 8️⃣ What is `seek()` and how is it used?
**Answer:** `seek(offset, whence)` moves the file cursor to a specific position.
```python
with open('example.txt', 'r') as file:
    file.seek(10)  # Moves to the 10th byte
    print(file.read())
```

---
### 9️⃣ How to delete a file in Python?
```python
import os
os.remove('example.txt')
```

---

with open('example.txt', 'r') as file:

        print(line.strip())
